# Section-A : Conceptual Understanding


A1. Why are NumPy arrays generally preferred over Python lists for numerical computation in AI/ML?

NumPy arrays are preferred because they are faster and more memory-efficient for numerical calculations. They support vectorized operations, so we can perform calculations on many values without writing explicit Python loops. NumPy also provides many mathematical and statistical functions that are useful in AI/ML.

```python
#A1 code:
import numpy as np 
    
arr = np.array([10,20,30])
print(arr * 2)
```

A2. Explain vectorization. Why does replacing an explicit Python loop with a vectorized NumPy operation matter with large datasets?

Vectorization means performing an operation on an entire NumPy array at once instead of processing each element using a Python loop. NumPy performs these operations efficiently using optimized internal code. This can make calculations much faster, especially when working with large datasets. It also makes the code shorter and easier to read.

```python
import numpy as np

x=np.array([1,2,3,4,5])
result = x * 10

print(result)
```

A3. For compatible 2D arrays A and B, explain A * B vs A @ B. Describe one neural-network situation for each.

"A * B" performs element-wise multiplication, meaning corresponding elements of the two arrays are multiplied. In a neural network, element-wise multiplication can be used for applying masks or element-wise operations on data. "A @ B" performs matrix multiplication and combines rows and columns according to matrix multiplication rules. It is commonly used to calculate the weighted sum between an input matrix and a weight matrix in a neural network.

```python
import numpy as np 

A = np.array([[1,2],
              [3,4]])

B = np.array([[5,6],
              [7,8]])

print("A * B:")
print(A * B)

print("A @ B:")
print(A @ B)
```

A4. What is broadcasting in NumPy?

Broadcasting is a NumPy feature that allows arrays with compatible different shapes to be used together in element-wise operations. NumPy compares their dimensions from the last dimension toward the first. Two dimensions are compatible when they are equal or when one of them is 1. This allows operations between arrays without manually changing their shapes.

```python
import numpy as np 

a = np.array([1,2,3])
b=10

print(a + b)
```

A5. Explain flatten() vs ravel(). In a memory-constrained application processing large images, which would you prefer and why?

"flatten()" converts an array into a 1D array and normally creates a copy of the data. "ravel()" also returns a 1D representation but tries to return a view of the original data when possible. Therefore, "ravel()" generally uses less additional memory. For a memory-constrained application processing large images, I would prefer "ravel()" when its view behavior is suitable.

```python
import numpy as np 

arr = np.array([[1,2],
                [3,4]])

print("Flatten:", arr.flatten())
print("Ravel:", arr.ravel())
```

A6. For a 2D array, explain axis=0 and axis=1 in an aggregation function.

For a 2D array, "axis=0" performs the operation down the rows, producing a result for each column. "axis=1" performs the operation across the columns, producing a result for each row. For example, for "[[1,2],[3,4]]", "np.sum(array, axis=0)" gives "[4,6]". "np.sum(array, axis=1)" gives "[3,7]"

```python
import numpy as np
arr = np.array([[1,2],
                [3,4]])

print("axis=0:", np.sum(arr, axis=0))
print("axis=1:", np.sum(arr, axis=1))
```

A7. Why is it important to normalize or standardize numerical features before training most ML models?

Normalization or standardization puts numerical features on comparable scales. This is important because features with very large values can have a greater influence on some ML algorithms. Normalization usually scales values to a fixed range such as 0 to 1. Standardization transforms values using the mean and standard deviation so that they are centered around 0 with a standard deviation of 1.


```python
import numpy as np

x = np.array([10, 20, 30, 40, 50])

normal = (x - x.min()) / (x.max() - x.min())
standard = (x - x.mean()) / x.std()

print("Normalization:",normal)
print("Standardization:",standard)
```

A8. Why do functions such as np.nanmean() exist?

"np.nanmean()" calculates the mean while ignoring "NaN" values. If "np.mean()" is directly applied to an array containing "NaN", the result will normally be "NaN". Therefore, "np.nanmean()" is useful when missing numerical values are present and we want to calculate the mean of the available values.

```python
import numpy as np

data = np.array([10, 20, np.nan, 40])

print("Mean:", np.mean(data))
print("NaN Mean:", np.nanmean(data))
```

A10. Why is reproducibility important during model development and experimentation?

Reproducibility means getting the same results when an experiment is repeated under the same conditions. Random operations can produce different results each time, which can make experiments difficult to compare. Using "np.random.default_rng(seed)" with a fixed seed makes the generated random values repeatable. This helps in debugging, testing, and comparing model experiments fairly.

```python
import numpy as np

rng1 = np.random.default_rng(10)
rng2 = np.random.default_rng(10)

a = rng1.random(5)
b = rng2.random(5)

print(a)
print(b)
print(np.array_equal(a, b))
```

# Section-D : Applied / Scenario Based - Problems

```python
# D1-A : Create data containing NaN

import numpy as np

data = np.array([
    [10.0, 20.0, np.nan],
    [30.0, 40.0, 50.0],
    [60.0, np.nan, 80.0]
])

print("Original data:")
print(data)
print("\n")


# B : Detect missing values

missing = np.isnan(data)

print("Missing value locations:")
print(missing)

print("Total missing values:", np.sum(missing))
print("\n")


# C : Handle missing values

column_mean = np.nanmean(data, axis=0)

rows, columns = np.where(np.isnan(data))

clean_data = data.copy()

clean_data[rows, columns] = column_mean[columns]

print("Clean data:")
print(clean_data)
print("\n")


# D : Standardize the clean data

mean = np.mean(clean_data, axis=0)
std = np.std(clean_data, axis=0)

standardized = (clean_data - mean) / std

print("Standardized data:")
print(standardized)
```

D2 – 1 RGB Image and Batch of RGB Images

An RGB image has 3 color channels: Red, Green, and Blue. A single image of size 64×64 pixels is represented as (64, 64, 3).
A batch of RGB images means multiple images are grouped together. For 16 images, the shape is (16, 64, 64, 3), where 16 represents the number of images.

```python
# D2 - 1: RGB Image and Batch RGB Images

# Single Image RGB

image_shape = (64, 64, 3)
print(image_shape)

# Batch Images RGB

batch_shape = (16, 64, 64, 3)

print(batch_shape)
print("\n")

# 2 : Creating the image batch

import numpy as np

rng = np.random.default_rng(10)

images = rng.integers(
    0, 256,
    size=(16, 64, 64, 3)
)

print("Batch shape:", images.shape)
print("\n")

# 3 : Calculate average image

average_image = np.mean(images, axis=0)

print("Average image shape:", average_image.shape)
```
